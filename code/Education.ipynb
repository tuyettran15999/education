## Rename columns

We will rename the columns to follow best practices of style and being informative. We can do this before or after joining the data sets. We will do it before joining the data sets to make it obvious that the key has the same name in each data set (this isn't required, but it makes your work easier to read).

```python
edgap = edgap.rename(
    columns={
        "NCESSCH School ID": "id",
        "CT Pct Adults with College Degree": "percent_college",
        "CT Unemployment Rate": "rate_unemployment",
        "CT Pct Childre In Married Couple Family": "percent_married",
        "CT Median Household Income": "median_income",
        "School ACT average (or equivalent if SAT score)": "average_act",
        "School Pct Free and Reduced Lunch": "percent_lunch"
    }
)
```

```python
school_information = school_information.rename(
    columns={
        'SCHOOL_YEAR': "year",
        'NCESSCH': "id",
        'LSTATE': "state",
        'LZIP': "zip_code",
        'SCH_TYPE_TEXT': "school_type",
        'LEVEL': "school_level",
        'CHARTER_TEXT': "charter"
    }
)
```

```python
school_geography = school_geography.rename(
    columns={
        'NCESSCH': "id",
        'LOCALE': "locale"
    }
)
```

```python
edgap.head()
```

```python
school_information.head()
```

```python
school_geography.head()
```

### Create locale_group for school geography data set

Based on Locale Definitions of National Center for Education Statistics (NCES), we create a locate_group column to get the geography of the school to see if it is in the city, suburban, town or rural.
Source: https://nces.ed.gov/surveys/annualreports/topical-studies/locale/definitions

```python
locale_map = {
    11: 'City',
    12: 'City',
    13: 'City',
    21: 'Suburban',
    22: 'Suburban',
    23: 'Suburban',
    31: 'Town',
    32: 'Town',
    33: 'Town',
    41: 'Rural',
    42: 'Rural',
    43: 'Rural'
}
```

```python
school_geography['locale'].dtype
```

```python
school_geography['locale'] = pd.to_numeric(
    school_geography['locale'],
    errors='coerce'
)
```

```python
school_geography['locale_group'] = (
    school_geography['locale'].map(locale_map)
)
```

```python
school_geography.head()
```

```python
school_geography.tail()
```

```python
school_geography['locale_group'].value_counts()
```

### Summary
1. We selected a subset of columns of the school information and school geography data sets.
2. We renamed the columns for clarity and to follow formatting guidelines.
3. We added a locale_group column for school geography data set.
4. We are ready to join the data frames.

## Join data frames

We want to join the three DataFrames using the NCES school identifier (NCESSCH) as the key. Because the identifier is stored using different data types across the datasets, we first standardize the ID formats to ensure accurate matching. The school geography IDs are converted from strings to integers, while the school information IDs are converted from floats to nullable integers to handle missing values, and then to objects to match the EdGap ID format.

```python
school_geography['id'] = school_geography['id'].astype(int)
```

```python
school_geography.info()
```

```python
school_information['id'] = school_information['id'].astype('Int64')
```

```python
school_information['id'] = school_information['id'].astype('object')
```

```python
school_information.info()
```

### Join the data frames and call the result df

```python
df = edgap.merge(school_information, how='left', on='id')
```

```python
df.head()
```

```python
edgap['id'].isin(school_geography['id']).value_counts()
```

```python
df = df.merge(school_geography, how='left', on='id')
```

```python
df.head()
```

```python
df.info()
```

### Summary
1. We joined the data sets using the school id as the key.
2. Before joining, we standardized the school ID data types across the datasets to ensure accurate matching.
3. We used a left join to keep all EdGap data.
4. After joining, the number of observations remained unchanged at 7,986.

## Data imputation

### Define the predictor variables to be rate_unemployment, percent_college, percent_married, median_income, percent_lunch, state, and charter.

```python
predictor_variables = ['rate_unemployment', 'percent_college', 'percent_married', 'median_income', 'percent_lunch', 'state','charter']
```

### Use the iterative imputer to replace missing values in the columns corresponding to predictor variables in the analysis.

```python
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer
```

```python
imputer = IterativeImputer()
```

### Fit the imputer using the numerical predictor variable (this can include dummies for categorical variables)

Define the columns you want to use in the imputation process.

```python
numerical_predictors = df[predictor_variables].select_dtypes(include='number').columns.to_list()
print(numerical_predictors)
```

### Fit the imputer

```python
imputer.fit(df.loc[:, numerical_predictors])
```

### Impute the missing values in the training data

```python
df.loc[:, numerical_predictors] = imputer.transform(df.loc[:, numerical_predictors])
```

### Check for the missing values

```python
df.isna().sum().to_frame(name='Number of Missing Values')
```

# Exploratory Data Analysis

## Examine distributions and relationships

### Plot the correlation matrix of the numerical variables in the training data to explore relationships between the variables.

```python
corr_matrix = df[numerical_predictors + ['average_act']].corr()

plt.figure(figsize=(10,5))

sns.heatmap(corr_matrix, vmax=1, vmin=-1, square=True, annot=True, cmap="viridis")

plt.tick_params(labelsize=12)

plt.show()
```

The correlation matrix shows that the percentage of students eligible for free or reduced-price lunch has the strongest relationship with average ACT scores (r=-0.78). College, median household income, and the percentage of children in married-couple families are moderately positively correlated with ACT scores, while unemployment is moderately negatively correlated with ACT scores.

### Make pair plots to explore relationships between the variables

```python
fig = sns.pairplot(data=df, vars=numerical_predictors + ['average_act'], hue='charter', kind="reg", plot_kws={"scatter_kws":{"alpha": 0.5 ,"color":"k","s":7}})

for ax in fig.axes.flat:
    if ax.get_xlabel() == 'CT Median Household Income':
        ax.ticklabel_format(style='sci', axis='x', scilimits=(0, 0)) # Apply scientific notation
    ax.set_xlabel(ax.get_xlabel(), fontsize=8, rotation=30, ha='right') # X-axis label size and rotation
    ax.set_ylabel(ax.get_ylabel(), fontsize=8)

    # Rotate x-axis tick labels
    plt.setp(ax.get_xticklabels(),rotation=30, ha='right')

plt.show()
```
